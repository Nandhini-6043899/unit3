```python
def kruskal(vertices, edges):
    parent = {v: v for v in vertices}

    def find(v):
        while parent[v] != v:
            parent[v] = parent[parent[v]]
            v = parent[v]
        return v

    def union(a, b):
        root_a = find(a)
        root_b = find(b)
        if root_a != root_b:
            parent[root_b] = root_a
            return True
        return False

    mst = []
    total = 0
    edges.sort(key=lambda x: x[2])

    for u, v, weight in edges:
        if union(u, v):
            mst.append((u, v, weight))
            total += weight

    return mst, total


vertices = ['A', 'B', 'C', 'D']
edges = [
    ('A', 'B', 4),
    ('A', 'C', 2),
    ('B', 'C', 1),
    ('B', 'D', 5),
    ('C', 'D', 8)
]

mst, total = kruskal(vertices, edges)

print("Minimum Spanning Tree:")
for edge in mst:
    print(edge)
print("Total Cost:", total)
```

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B'],
    'F': ['C']
}


def bfs(start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


def dfs(vertex, visited=None):
    if visited is None:
        visited = set()

    visited.add(vertex)
    print(vertex, end=" ")
    for neighbor in graph[vertex]:
        if neighbor not in visited:
            dfs(neighbor, visited)


print("BFS:")
bfs('A')
print("\nDFS:")
dfs('A')
```

```python
"""
LAB 2: Heap Implementation (Max Heap, array based)
Operations: 1. Heap creation   2. Insertion   3. Deletion
            4. Display         5. Heapify

For a node at index i:   parent = (i - 1) // 2,   left = 2i + 1,   right = 2i + 2
"""


class MaxHeap:
    def __init__(self):
        self.heap = []

    # ---------- helper functions ----------
    def parent(self, i):
        return (i - 1) // 2

    def left(self, i):
        return 2 * i + 1

    def right(self, i):
        return 2 * i + 2

    # ---------- 5. HEAPIFY ----------
    def heapify_up(self, i):
        """Move a node up until its parent is larger (used after insertion)."""
        while i > 0 and self.heap[self.parent(i)] < self.heap[i]:
            p = self.parent(i)
            self.heap[i], self.heap[p] = self.heap[p], self.heap[i]
            i = p

    def heapify_down(self, i, size=None):
        """Move a node down until both children are smaller (used after deletion)."""
        size = len(self.heap) if size is None else size
        while True:
            largest = i
            l, r = self.left(i), self.right(i)
            if l < size and self.heap[l] > self.heap[largest]:
                largest = l
            if r < size and self.heap[r] > self.heap[largest]:
                largest = r
            if largest == i:
                break
            self.heap[i], self.heap[largest] = self.heap[largest], self.heap[i]
            i = largest

    def build_heap(self, array):
        """1. HEAP CREATION from an unordered array in O(n) using heapify."""
        self.heap = list(array)
        for i in range(len(self.heap) // 2 - 1, -1, -1):     # last parent up to root
            self.heapify_down(i)

    # ---------- 2. INSERTION ----------
    def insert(self, value):
        self.heap.append(value)                 # add at the end
        self.heapify_up(len(self.heap) - 1)     # restore the heap property

    # ---------- 3. DELETION ----------
    def delete_root(self):
        """Remove and return the largest element (the root)."""
        if not self.heap:
            print("Heap is empty")
            return None
        top = self.heap[0]
        last = self.heap.pop()
        if self.heap:
            self.heap[0] = last                 # move the last element to the root
            self.heapify_down(0)                # push it down to its place
        return top

    def delete_value(self, value):
        """Delete any given value from the heap."""
        if value not in self.heap:
            print(f"{value} not found in the heap")
            return False
        i = self.heap.index(value)
        last = self.heap.pop()
        if i < len(self.heap):
            self.heap[i] = last
            self.heapify_up(i)
            self.heapify_down(i)
        return True

    # ---------- 4. DISPLAY ----------
    def display(self):
        print("  Array form :", self.heap)
        if not self.heap:
            return
        print("  Tree form  :")
        n = len(self.heap)
        level, start = 0, 0
        while start < n:
            end = min(start + 2 ** level, n)
            print(f"    Level {level}: {self.heap[start:end]}")
            start, level = end, level + 1

    def is_heap(self):
        return all(self.heap[self.parent(i)] >= self.heap[i] for i in range(1, len(self.heap)))


def title(text):
    print("\n" + "=" * 60)
    print(text)
    print("=" * 60)


# ===============================================================
# 1. HEAP CREATION
# ===============================================================
title("1. HEAP CREATION")
data = [10, 20, 15, 30, 40, 50, 5]
print("  Unordered input :", data)

h = MaxHeap()
h.build_heap(data)
print("  After build_heap (bottom-up heapify):")
h.display()
print("  Valid max heap? ", h.is_heap())

# ===============================================================
# 2. INSERTION
# ===============================================================
title("2. INSERTION")
for value in (60, 25):
    h.insert(value)
    print(f"  Inserted {value}:")
    h.display()
print("  Valid max heap? ", h.is_heap())

# ===============================================================
# 3. DELETION
# ===============================================================
title("3. DELETION")
removed = h.delete_root()
print(f"  Deleted root (maximum) = {removed}")
h.display()

print("\n  Delete the value 15 from the middle of the heap:")
h.delete_value(15)
h.display()
print("  Valid max heap? ", h.is_heap())

# ===============================================================
# 4. DISPLAY
# ===============================================================
title("4. DISPLAY")
h.display()

# ===============================================================
# 5. HEAPIFY (one step shown, then repeated deletion = heap sort)
# ===============================================================
title("5. HEAPIFY")
broken = MaxHeap()
broken.heap = [5, 40, 30, 10, 20, 15, 25]
print("  Array where only the root breaks the heap rule:", broken.heap)
broken.heapify_down(0)
print("  After heapify_down(0):", broken.heap)

print("\n  Heap sort: delete the root repeatedly to get descending order:")
sorter = MaxHeap()
sorter.build_heap([12, 3, 19, 8, 25, 1])
result = []
while sorter.heap:
    result.append(sorter.delete_root())
print("  Sorted (descending):", result)

print("\nTime complexity: insert O(log n), delete O(log n), build_heap O(n), peek O(1)")
```

```python
"""
LAB 1: Binary Tree and Expression Tree
Part A: Binary tree with inorder, preorder and postorder traversals
Part B: Expression tree: construct, traverse and evaluate  (5 + 3) * 2
"""


class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# ---------------------------------------------------------------
# Traversals (work for any binary tree)
# ---------------------------------------------------------------
def inorder(node, result):        # Left, Root, Right
    if node:
        inorder(node.left, result)
        result.append(node.data)
        inorder(node.right, result)
    return result


def preorder(node, result):       # Root, Left, Right
    if node:
        result.append(node.data)
        preorder(node.left, result)
        preorder(node.right, result)
    return result


def postorder(node, result):      # Left, Right, Root
    if node:
        postorder(node.left, result)
        postorder(node.right, result)
        result.append(node.data)
    return result


def print_tree(node, level=0, label="root"):
    """Sideways view of the tree: right subtree on top, left at the bottom."""
    if node is None:
        return
    print_tree(node.right, level + 1, "R")
    print("      " * level + f"[{label}] {node.data}")
    print_tree(node.left, level + 1, "L")


# ===============================================================
# PART A: Binary tree and its three traversals
# ===============================================================
print("=" * 60)
print("PART A: BINARY TREE AND TRAVERSALS")
print("=" * 60)

#            10
#          /    \
#        20      30
#       /  \       \
#     40    50      60
root = Node(10)
root.left = Node(20)
root.right = Node(30)
root.left.left = Node(40)
root.left.right = Node(50)
root.right.right = Node(60)

print("\nTree (rotated 90 degrees, right side on top):")
print_tree(root)

print("\n1. Inorder   (Left, Root, Right):", inorder(root, []))
print("2. Preorder  (Root, Left, Right):", preorder(root, []))
print("3. Postorder (Left, Right, Root):", postorder(root, []))


# ===============================================================
# PART B: Expression tree
# ===============================================================
PRECEDENCE = {"+": 1, "-": 1, "*": 2, "/": 2, "^": 3}


def tokenize(expression):
    tokens, number = [], ""
    for ch in expression:
        if ch.isdigit():
            number += ch
        else:
            if number:
                tokens.append(number)
                number = ""
            if ch != " ":
                tokens.append(ch)
    if number:
        tokens.append(number)
    return tokens


def infix_to_postfix(expression):
    """Shunting-yard algorithm."""
    output, stack = [], []
    for tok in tokenize(expression):
        if tok.isdigit():
            output.append(tok)
        elif tok == "(":
            stack.append(tok)
        elif tok == ")":
            while stack and stack[-1] != "(":
                output.append(stack.pop())
            stack.pop()                                   # discard "("
        else:
            while (stack and stack[-1] != "(" and
                   (PRECEDENCE[stack[-1]] > PRECEDENCE[tok] or
                    (PRECEDENCE[stack[-1]] == PRECEDENCE[tok] and tok != "^"))):
                output.append(stack.pop())
            stack.append(tok)
    while stack:
        output.append(stack.pop())
    return output


def build_expression_tree(postfix_tokens):
    """Operand -> push a leaf. Operator -> pop two subtrees, make them its children."""
    stack = []
    for tok in postfix_tokens:
        node = Node(tok)
        if tok.isdigit():
            stack.append(node)
        else:
            node.right = stack.pop()
            node.left = stack.pop()
            stack.append(node)
    return stack.pop()


def evaluate(node):
    """Evaluate the tree recursively (postorder idea)."""
    if node.left is None and node.right is None:
        return int(node.data)
    left_value = evaluate(node.left)
    right_value = evaluate(node.right)
    if node.data == "+":
        return left_value + right_value
    if node.data == "-":
        return left_value - right_value
    if node.data == "*":
        return left_value * right_value
    if node.data == "/":
        return left_value / right_value
    if node.data == "^":
        return left_value ** right_value
    raise ValueError("Unknown operator " + node.data)


def infix_with_brackets(node):
    """Inorder traversal that restores the brackets."""
    if node.left is None and node.right is None:
        return node.data
    return "(" + infix_with_brackets(node.left) + " " + node.data + " " + \
           infix_with_brackets(node.right) + ")"


print("\n" + "=" * 60)
print("PART B: EXPRESSION TREE")
print("=" * 60)

expression = "(5 + 3) * 2"
print("\nInfix expression :", expression)

postfix = infix_to_postfix(expression)
print("Converted postfix:", " ".join(postfix))

tree = build_expression_tree(postfix)
print("\n4. Expression tree (rotated 90 degrees, right side on top):")
print_tree(tree)
print("""
Same tree drawn normally:
        *
       / \\
      +   2
     / \\
    5   3""")

print("\n5. Traversals of the expression tree:")
print("   Inorder   (infix)  :", " ".join(inorder(tree, [])), "  ->  with brackets:", infix_with_brackets(tree))
print("   Preorder  (prefix) :", " ".join(preorder(tree, [])))
print("   Postorder (postfix):", " ".join(postorder(tree, [])))

print("\n6. Evaluation of the expression:")
print("   Result =", evaluate(tree))

# More examples to show the same code works for other expressions
print("\nMore examples:")
for exp in ("2 + 3 * 4", "(10 - 4) / 3", "((2 + 3) * (8 - 4)) ^ 2"):
    t = build_expression_tree(infix_to_postfix(exp))
    print(f"   {exp:<26} postfix: {' '.join(postorder(t, [])):<18} result = {evaluate(t)}")
```

```python
# Q12: Construct a binary tree; find total, leaf and internal nodes
from collections import deque


class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    def insert(self, data):
        """Insert at the first vacant position, scanning level by level."""
        new_node = Node(data)
        if self.root is None:
            self.root = new_node
            return
        queue = deque([self.root])
        while queue:
            node = queue.popleft()
            if node.left is None:
                node.left = new_node
                return
            queue.append(node.left)
            if node.right is None:
                node.right = new_node
                return
            queue.append(node.right)

    def inorder(self, node, result):
        if node:
            self.inorder(node.left, result)
            result.append(node.data)
            self.inorder(node.right, result)
        return result

    def count_total(self, node):
        if node is None:
            return 0
        return 1 + self.count_total(node.left) + self.count_total(node.right)

    def count_leaf(self, node):
        if node is None:
            return 0
        if node.left is None and node.right is None:
            return 1
        return self.count_leaf(node.left) + self.count_leaf(node.right)

    def count_internal(self, node):
        if node is None or (node.left is None and node.right is None):
            return 0
        return 1 + self.count_internal(node.left) + self.count_internal(node.right)

    def height(self, node):
        """Height counted in levels (an empty tree has height 0)."""
        if node is None:
            return 0
        return 1 + max(self.height(node.left), self.height(node.right))

    def show_levels(self):
        """Print the tree level by level."""
        if self.root is None:
            print("Tree is empty")
            return
        queue = deque([(self.root, 1)])
        current, line = 1, []
        while queue:
            node, level = queue.popleft()
            if level != current:
                print(f"Level {current}: {line}")
                current, line = level, []
            line.append(node.data)
            if node.left:
                queue.append((node.left, level + 1))
            if node.right:
                queue.append((node.right, level + 1))
        print(f"Level {current}: {line}")


tree = BinaryTree()
for value in (10, 20, 30, 40, 50, 60, 70, 80):
    tree.insert(value)

print("Tree, level by level:")
tree.show_levels()

total = tree.count_total(tree.root)
leaves = tree.count_leaf(tree.root)
internal = tree.count_internal(tree.root)

print("\nInorder traversal :", tree.inorder(tree.root, []))
print("Total nodes       :", total)
print("Leaf nodes        :", leaves)
print("Internal nodes    :", internal)
print("Check: leaf + internal == total ->", leaves + internal == total)
```

```python
# Q11: Stack using a list with Push, Pop, Peek, Display and IsEmpty
class Stack:
    """Stack using a Python list (top of the stack = end of the list)."""

    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, value):
        self.items.append(value)
        print(f"Push({value})  ->  stack (bottom to top): {self.items}")

    def pop(self):
        if self.is_empty():
            print("Stack Underflow: cannot pop from an empty stack")
            return None
        value = self.items.pop()
        print(f"Pop() -> {value}  ->  stack (bottom to top): {self.items}")
        return value

    def peek(self):
        return None if self.is_empty() else self.items[-1]

    def display(self):
        print("Stack (bottom to top):", self.items if self.items else "empty")


s = Stack()

print("Test 1: new stack")
print("IsEmpty?", s.is_empty())
s.pop()

print("\nTest 2: push 10, 20, 30, 40")
for value in (10, 20, 30, 40):
    s.push(value)
s.display()
print("Peek:", s.peek())

print("\nTest 3: pop twice")
s.pop()
s.pop()
s.display()
print("Peek:", s.peek())
print("IsEmpty?", s.is_empty())

print("\nTest 4: pop the rest")
s.pop()
s.pop()
s.display()
print("IsEmpty?", s.is_empty())
```

```python
# Q10: Role of the stack in function calls and recursion (call stack demo)
call_stack = []


def factorial(n):
    call_stack.append(f"factorial({n})")                 # push a frame
    print("  " * (len(call_stack) - 1) + f"CALL   factorial({n})    call stack: {call_stack}")
    result = 1 if n <= 1 else n * factorial(n - 1)
    frame = call_stack.pop()                             # pop the frame
    print("  " * len(call_stack) + f"RETURN {frame} = {result}")
    return result


print("Computing factorial(4); watch frames being pushed, then popped:\n")
answer = factorial(4)
print("\nFinal answer:", answer)
print("Call stack is now:", call_stack)

# Without a base case the call stack keeps growing until it overflows.
def endless(n):
    return endless(n + 1)

try:
    endless(0)
except RecursionError:
    print("\nRecursion with no base case -> RecursionError (stack overflow)")
```

```python
# Q8: Structure of a binary tree: root, parent, leaf and internal nodes

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


root = Node("A")
root.left, root.right = Node("B"), Node("C")
root.left.left, root.left.right = Node("D"), Node("E")
root.right.left = Node("F")


def classify(node, parent=None):
    if node is None:
        return
    is_leaf = node.left is None and node.right is None
    roles = []
    if parent is None:
        roles.append("ROOT")
    roles.append("LEAF" if is_leaf else "INTERNAL")
    if not is_leaf:
        children = [c.data for c in (node.left, node.right) if c]
        roles.append("PARENT of " + ", ".join(children))
    print(f"{node.data}:  " + " | ".join(roles) + (f"   (its parent is {parent})" if parent else ""))
    classify(node.left, node.data)
    classify(node.right, node.data)


classify(root)

leaves = sum(1 for n in "DEF")
print("\nLeaf nodes: 3 (D, E, F)   Internal nodes: 3 (A, B, C)   Total: 6")
```

```python
# Q6: Evaluation of a postfix expression using a stack
def evaluate_postfix(expression):
    stack = []
    for token in expression.split():
        if token.lstrip("-").isdigit():
            stack.append(int(token))                  # operand: push
        else:
            right = stack.pop()                       # operator: pop two operands
            left = stack.pop()
            if token == "+":
                result = left + right
            elif token == "-":
                result = left - right
            elif token == "*":
                result = left * right
            elif token == "/":
                result = left / right
            elif token == "^":
                result = left ** right
            else:
                raise ValueError(f"Unknown operator: {token}")
            stack.append(result)                      # push the result back
        print(f"token {token:>2}   ->   stack: {stack}")
    return stack.pop()


for expr, infix in (("5 3 + 8 2 - *", "(5 + 3) * (8 - 2)"),
                    ("2 3 1 * + 9 -", "2 + 3 * 1 - 9")):
    print(f"\nPostfix: {expr}      Infix: {infix}")
    print("Result =", evaluate_postfix(expr))
```

```python
# Q5: Push and Pop with an example, and the time complexity of each
import time
class Stack:
    """Stack using a Python list (top of the stack = end of the list)."""

    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, value):
        self.items.append(value)
        print(f"Push({value})  ->  stack (bottom to top): {self.items}")

    def pop(self):
        if self.is_empty():
            print("Stack Underflow: cannot pop from an empty stack")
            return None
        value = self.items.pop()
        print(f"Pop() -> {value}  ->  stack (bottom to top): {self.items}")
        return value

    def peek(self):
        return None if self.is_empty() else self.items[-1]

    def display(self):
        print("Stack (bottom to top):", self.items if self.items else "empty")


print("Example: Push(7), Push(9), Pop(), Push(4)")
s = Stack()
s.push(7)
s.push(9)
s.pop()
s.push(4)

print("\nTime complexity: Push = O(1), Pop = O(1)")
print("Both operations only touch the top element, so the cost does not")
print("grow with the stack size. Timing check (time per operation):\n")


class QuietStack:
    def __init__(self):
        self.items = []
    def push(self, v):
        self.items.append(v)
    def pop(self):
        return self.items.pop()


for n in (10_000, 100_000, 1_000_000):
    q = QuietStack()
    t0 = time.perf_counter()
    for i in range(n):
        q.push(i)
    push_time = (time.perf_counter() - t0) / n
    t0 = time.perf_counter()
    for _ in range(n):
        q.pop()
    pop_time = (time.perf_counter() - t0) / n
    print(f"n = {n:>9,}   push: {push_time*1e9:6.0f} ns/op   pop: {pop_time*1e9:6.0f} ns/op")
```

```python
# Q4: Insertion of nodes into a binary tree (level-order insertion)
from collections import deque


class Node1:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinaryTree:
    def __init__(self):
        self.root = None

    def insert(self, data):
        """Insert at the first vacant position, scanning level by level."""
        new_node = Node(data)
        if self.root is None:
            self.root = new_node
            return
        queue = deque([self.root])
        while queue:
            node = queue.popleft()
            if node.left is None:
                node.left = new_node
                return
            queue.append(node.left)
            if node.right is None:
                node.right = new_node
                return
            queue.append(node.right)

    def inorder(self, node, result):
        if node:
            self.inorder(node.left, result)
            result.append(node.data)
            self.inorder(node.right, result)
        return result

    def count_total(self, node):
        if node is None:
            return 0
        return 1 + self.count_total(node.left) + self.count_total(node.right)

    def count_leaf(self, node):
        if node is None:
            return 0
        if node.left is None and node.right is None:
            return 1
        return self.count_leaf(node.left) + self.count_leaf(node.right)

    def count_internal(self, node):
        if node is None or (node.left is None and node.right is None):
            return 0
        return 1 + self.count_internal(node.left) + self.count_internal(node.right)

    def height(self, node):
        """Height counted in levels (an empty tree has height 0)."""
        if node is None:
            return 0
        return 1 + max(self.height(node.left), self.height(node.right))

    def show_levels(self):
        """Print the tree level by level."""
        if self.root is None:
            print("Tree is empty")
            return
        queue = deque([(self.root, 1)])
        current, line = 1, []
        while queue:
            node, level = queue.popleft()
            if level != current:
                print(f"Level {current}: {line}")
                current, line = level, []
            line.append(node.data)
            if node.left:
                queue.append((node.left, level + 1))
            if node.right:
                queue.append((node.right, level + 1))
        print(f"Level {current}: {line}")


tree = BinaryTree()
for value in (10, 20, 30, 40, 50):
    tree.insert(value)
    print(f"Inserted {value:>2}  ->  inorder now: {tree.inorder(tree.root, [])}")

print("\nFinal tree, level by level:")
tree.show_levels()
print("\nRoot            :", tree.root.data)
print("Root's children :", tree.root.left.data, "and", tree.root.right.data)
print("20's children   :", tree.root.left.left.data, "and", tree.root.left.right.data)
```

```python
# Q3: Inorder traversal of a binary tree with 10, 20, 30, 40, 50

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def inorder(node, result):
    """Left -> Root -> Right"""
    if node:
        inorder(node.left, result)
        result.append(node.data)
        inorder(node.right, result)
    return result


root = Node(10)
root.left = Node(20)
root.right = Node(30)
root.left.left = Node(40)
root.left.right = Node(50)

print("Inorder traversal (Left, Root, Right):")
print(inorder(root, []))
```
